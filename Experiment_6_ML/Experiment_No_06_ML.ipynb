# Import Required Libraries

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

# Load Iris Dataset

```python
df = sns.load_dataset("iris")

df.head()
```

```python
print("Shape of Dataset:", df.shape)

print("\nDataset Information:")
df.info()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
df.describe()
```

```python
print(df["species"].value_counts())
```

# Visualize the Dataset

```python
sns.pairplot(df, hue="species")

plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    df.drop("species", axis=1).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```

# Decision Tree Classification

```python
X = df.drop("species", axis=1)
y = df["species"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

# Decision Tree Model

```python
dt_model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)
```

# Train Decision Tree Model

```python
dt_model.fit(X_train, y_train)

print("Decision Tree Model Trained Successfully!")
```

```python
y_pred_dt = dt_model.predict(X_test)

print("Predicted Values:")
print(y_pred_dt)
```

# Calculate Decision Tree Accuracy

```python
dt_accuracy = accuracy_score(y_test, y_pred_dt)

print("Decision Tree Accuracy:", dt_accuracy)
print("Decision Tree Accuracy (%):", dt_accuracy * 100)
```

# Decision Tree Confusion Matrix

```python
cm_dt = confusion_matrix(y_test, y_pred_dt)

print("Decision Tree Confusion Matrix:")
print(cm_dt)
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_dt,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=dt_model.classes_,
    yticklabels=dt_model.classes_
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("Decision Tree Confusion Matrix")

plt.show()
```

# Visualize the Decision Tree

```python
from sklearn.tree import plot_tree

plt.figure(figsize=(20, 10))

plot_tree(
    dt_model,
    feature_names=X.columns,
    class_names=dt_model.classes_,
    filled=True
)

plt.title("Decision Tree for Iris Classification")

plt.show()
```

# K-Nearest Neighbour (KNN) Classification

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("Feature Scaling Completed!")
```

# Create KNN Model

```python
knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train_scaled, y_train)

print("KNN Model Trained Successfully!")

```

```python
y_pred_knn = knn_model.predict(X_test_scaled)

print("KNN Predicted Values:")
print(y_pred_knn)
```

# KNN Accuracy

```python
knn_accuracy = accuracy_score(y_test, y_pred_knn)

print("KNN Accuracy:", knn_accuracy)
print("KNN Accuracy (%):", knn_accuracy * 100)
```

# KNN Confusion Matrix

```python
cm_knn = confusion_matrix(y_test, y_pred_knn)

print("KNN Confusion Matrix:")
print(cm_knn)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm_knn,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=knn_model.classes_,
    yticklabels=knn_model.classes_
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("KNN Confusion Matrix")

plt.show()
```

# Compare Decision Tree and KNN

```python
print("Decision Tree Accuracy:", dt_accuracy)
print("KNN Accuracy:", knn_accuracy)
```

# Comparison DataFrame

```python
results = pd.DataFrame({
    "Algorithm": ["Decision Tree", "KNN"],
    "Accuracy": [dt_accuracy, knn_accuracy]
})

results
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=results,
    x="Algorithm",
    y="Accuracy"
)

plt.ylim(0, 1.1)

plt.title("Decision Tree vs KNN Accuracy")
plt.xlabel("Algorithm")
plt.ylabel("Accuracy")

plt.show()
```
