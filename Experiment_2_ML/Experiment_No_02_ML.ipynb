# 7. More Analysis

```python
sns.countplot(data=titanic, x='survived', palette='autumn', hue='sex')
```

```python
sns.countplot(data=titanic, x='survived', palette='viridis', hue='class')
```

```python
titanic.age.plot(kind='hist',bins=30,color='green')
```

```python
sns.scatterplot(data=titanic, x='age', y='fare', hue='class')
```

```python
sns.scatterplot(data=titanic, x='age', y='fare', hue='sex')
```

```python
# Survival Rate by Passenger Class
survival_class = titanic.groupby('pclass')['survived'].mean()
print(survival_class)

sns.barplot(data=titanic, x='pclass', y='survived')
plt.title('Survival Rate by Passenger Class')
plt.show()
```

```python
# Survival Rate by Gender
survival_sex = titanic.groupby('sex')['survived'].mean()
print(survival_sex)

sns.barplot(data=titanic, x='sex', y='survived')
plt.title('Survival Rate by Gender')
plt.show()
```

```python
# Survival by Gender and Class
sns.barplot(data=titanic, x='sex', y='survived', hue='class')
plt.title('Survival by Gender and Class')
plt.show()
```

```python
# Boxplot for Age
sns.boxplot(data=titanic, x='class', y='age')
plt.title('Age Distribution by Class')
plt.show()
```

```python
# Age vs Survival
sns.boxplot(data=titanic, x='survived', y='age')
plt.title('Age vs Survival')
plt.show()
```
