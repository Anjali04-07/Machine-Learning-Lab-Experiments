# Import required libraries

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

# Load Iris dataset

```python
df = sns.load_dataset("iris")

df.head()
```

# Display dataset information

```python
print("Shape of dataset:", df.shape)

print("\nDataset Information:")
df.info()
```

```python
df.describe()
```

```python
print(df.isnull().sum())
```

```python
print(df["species"].value_counts())
```

```python
sns.pairplot(df, hue="species")
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    df.drop(columns=["species"]).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

# Separate input and output

```python
X = df.drop("species", axis=1)
y = df["species"]

print("Input features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

# Split dataset into training and testing data

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

# Feature Scaling

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# Create SVM model

```python
model = SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale"
)
```

# Train the SVM model

```python
model.fit(X_train_scaled, y_train)
```

```python
y_pred = model.predict(X_test_scaled)

print("Predicted values:")
print(y_pred)
```

# Compare actual and predicted values

```python
comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

comparison
```

# Calculate accuracy

```python
accuracy = accuracy_score(y_test, y_pred)

print("SVM Accuracy:", accuracy)
print("SVM Accuracy (%):", accuracy * 100)
```

# Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)


plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=model.classes_,
    yticklabels=model.classes_
)

plt.xlabel("Predicted Label")
plt.ylabel("Actual Label")
plt.title("SVM Confusion Matrix")

plt.show()
```

```python
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

# Test SVM with different kernels

```python
kernels = ["linear", "poly", "rbf", "sigmoid"]

for kernel in kernels:
    svm_model = SVC(kernel=kernel)
    
    svm_model.fit(X_train_scaled, y_train)
    
    prediction = svm_model.predict(X_test_scaled)
    
    accuracy = accuracy_score(y_test, prediction)
    
    print(kernel, "Kernel Accuracy:", accuracy)
```

# Plot kernel accuracy using Seaborn

```python
results = []

for kernel in kernels:
    svm_model = SVC(kernel=kernel)
    svm_model.fit(X_train_scaled, y_train)
    
    prediction = svm_model.predict(X_test_scaled)
    
    accuracy = accuracy_score(y_test, prediction)
    
    results.append({
        "Kernel": kernel,
        "Accuracy": accuracy
    })

results_df = pd.DataFrame(results)

results_df
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=results_df,
    x="Kernel",
    y="Accuracy"
)

plt.ylim(0, 1.1)
plt.title("SVM Accuracy for Different Kernels")
plt.ylabel("Accuracy")
plt.xlabel("Kernel")

plt.show()
```
